```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

def run_eda():
    print("Starting Exploratory Data Analysis...")
    
    # 1. Load the data
    data_path = os.path.join("data", "raw", "SP500_DATASET.csv")
    
    if not os.path.exists(data_path):
        print(f"ERROR: Dataset not found at {data_path}. Please place your CSV file there.")
        return
        
    df = pd.read_csv(data_path, parse_dates=['Date'], index_col='Date')
    df.dropna(inplace=True)
    print(f"Data loaded successfully. Shape: {df.shape}")

    # Set plot styles
    plt.style.use('seaborn-v0_8-darkgrid')
    sns.set_theme(style='darkgrid')

    # 2. Plot 1: Historical Closing Prices
    print("Generating Historical Prices plot... (Close the window to continue)")
    plt.figure(figsize=(14, 6))
    plt.plot(df.index, df['Close'], label='S&P 500 Close Price', color='blue')
    plt.title('S&P 500 Historical Closing Prices')
    plt.xlabel('Date')
    plt.ylabel('Price (USD)')
    plt.legend()
    plt.show() # Code pauses here until you close the plot window

    # 3. Engineer Features
    print("Engineering Technical Indicators...")
    df['MA_20'] = df['Close'].rolling(window=20).mean()
    df['MA_50'] = df['Close'].rolling(window=50).mean()
    df['Volatility_20'] = df['Close'].rolling(window=20).std()
    df['Daily_Return'] = df['Close'].pct_change()

    delta = df['Close'].diff(1)
    gain = (delta.where(delta > 0, 0)).rolling(window=14).mean()
    loss = (-delta.where(delta < 0, 0)).rolling(window=14).mean()
    rs = gain / loss
    df['RSI_14'] = 100 - (100 / (1 + rs))

    df.dropna(inplace=True)

    # 4. Plot 2: Moving Averages (Last 500 days)
    print("Generating Moving Averages plot... (Close the window to continue)")
    plt.figure(figsize=(14, 6))
    plt.plot(df.index[-500:], df['Close'].iloc[-500:], label='Close', alpha=0.5)
    plt.plot(df.index[-500:], df['MA_20'].iloc[-500:], label='20-Day MA', color='red')
    plt.plot(df.index[-500:], df['MA_50'].iloc[-500:], label='50-Day MA', color='green')
    plt.title('S&P 500 with Moving Averages (Last 500 Days)')
    plt.legend()
    plt.show()

    # 5. Plot 3: Correlation Heatmap
    print("Generating Correlation Heatmap... (Close the window to finish)")
    plt.figure(figsize=(10, 8))
    features = ['Open', 'High', 'Low', 'Close', 'Volume', 'MA_20', 'MA_50', 'Volatility_20', 'RSI_14']
    sns.heatmap(df[features].corr(), annot=True, cmap='coolwarm', fmt='.2f')
    plt.title('Feature Correlation Heatmap')
    plt.show()
    
    print("EDA Complete!")

if __name__ == "__main__":
    run_eda()
```
